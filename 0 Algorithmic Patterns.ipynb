## Two Pointers

The Two Pointers pattern utilizes two index variables to traverse a linear data structure (like an array or a string) simultaneously. It generally reduces an O(N²) brute-force solution to an optimal O(N) time complexity by eliminating unnecessary nested loops.

**When to use**: Working with sorted arrays or strings where you need to find pairs or compare elements from opposite ends.

**Classic Problems**: Pair with Target Sum (Two Sum II), Valid Palindrome, Container With Most Water.

```python
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1  # Need a larger sum
        else:
            right -= 1 # Need a smaller sum
            
    return [-1, -1]

two_sum_sorted([1, 2, 3, 4], 4)
```

## Sliding Window

The Sliding Window pattern involves maintaining a sub-segment of data (the "window") over a longer array or string. Instead of recalculating the entire window from scratch as it moves, you dynamically track elements entering from the right and leaving from the left.

**When to use**: Problems involving contiguous subarrays or substrings where you need to track a running max, min, or distinct count.

**Classic Problems**: Maximum Sum Subarray of Size K, Longest Substring Without Repeating Characters.

```python
def longest_substring_without_repeats(s):
    char_set = set()
    left = 0
    max_length = 0
    
    for right in range(len(s)):
        # Shrink window from left until the duplicate character is removed
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        char_set.add(s[right])
        max_length = max(max_length, right - left + 1)
    return max_length

longest_substring_without_repeats(["h", "e", "l", "l", "o"])
```

## Prefix Sum

The Prefix Sum pattern involves pre-computing an array of cumulative sums up to each index. This allows you to answer any subarray or range-sum query in O(1) constant time instead of running an O(N) loop every time.

**When to use**: Multiple queries tracking the sum of elements within a specific subarray or range.

**Classic Problems**: Range Sum Query (Immutable), Subarray Sum Equals K, Product of Array Except Self.

```python
class PrefixSum:
    def __init__(self, arr):
        # Pad with a leading 0 to easily handle queries starting at index 0
        self.prefix = [0] * (len(arr) + 1)
        for i in range(len(arr)):
            self.prefix[i + 1] = self.prefix[i] + arr[i]
    
    def query_range(self, left, right):
        # Returns sum of arr[left...right] inclusive
        return self.prefix[right + 1] - self.prefix[left]
```

## Fast & Slow Pointers (Tortoise and Hare)

This pattern uses two pointers moving through a linear data structure (usually a Linked List or an array) at different speeds. The fast pointer typically moves two steps at a time, while the slow pointer moves one step. If a cycle exists, the pointers are guaranteed to meet.

**When to use**: Detecting cycles in a linked list or array, or finding the exact midpoint of a structural chain.

**Classic Problems**: Linked List Cycle, Find the Duplicate Number, Middle of the Linked List.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
def has_cycle(head):
    slow, fast = head, head
    
    while fast and fast.next:
        slow = slow.next      # Moves 1 step
        fast = fast.next.next # Moves 2 steps
        
        if slow == fast:      # Fast pointer caught up: cycle detected
            return True
    
    return False
```

## Top 'K' Elements (Heaps)

Finding the top, smallest, or most frequent K elements in an unsorted dataset normally takes O(NlogN) sorting time. Using a Min-Heap or Max-Heap allows you to maintain the top K items dynamically in O(NlogK) time, which is highly optimal when K << N.

**When to use**: Asked to find the "K largest", "K smallest", or "K most frequent" items.

**Classic Problems**: Kth Largest Element in an Array, Top K Frequent Elements, Merge K Sorted Lists.

```python
import heapq

def find_k_largest(nums, k):
    # Initialize a min-heap with the first K elements
    min_heap = nums[:k]
    heapq.heapify(min_heap) # O(K) time to structure the heap
    
    # Process remaining elements
    for num in nums[k:]:
        if num > min_heap[0]:
            heapq.heappushpop(min_heap, num) # Keeps heap size exactly K
    
    return min_heap # Contains the K largest elements in the array

find_k_largest([1, 3, 5, 5, 7, 8], 3)
```

## Depth-First Search (DFS)

**Problem**: Max Depth of a Binary Tree (LeetCode 104)

**Strategy**: Break the problem down recursively. The height of any node is 1 plus the maximum height of its left and right subtrees. Return 0 as the base case when a null node is reached.

```python
def maxDepth(root):
    # Base case - an empry node contributes 0 to the depth
    if not root:
        return 0
    
    # Recursive case - find depths of subtrees
    left_depth = maxDepth(root.left)
    right_depth = maxDepth(root.right)
    
    return max(left_depth, right_depth) + 1
```

**How it executes step-by-step**:

Given a tree where 3 links to left child 9 (leaf) and right child 20 (has leaf children):
- maxDepth(3) calls maxDepth(9) and maxDepth(20).
- maxDepth(9) looks at its empty children, receives 0 and 0, and returns 0 + 1 = 1.
- maxDepth(20) calls its children 15 and 7. Both return 1. maxDepth(20) returns max(1, 1) + 1 = 2.
- maxDepth(3) evaluates max(1, 2) + 1 and returns 3.

## Backtracking

**Problem**: Subsets / Power Set (LeetCode 78)

**Strategy**: Decide for every single element whether to include it or exclude it. Choose the element, explore further down the decision tree via recursion, and then un-choose (backtrack) to explore the alternative path.

```python
def subsets(nums): # Defines the main function that takes a list of numbers (e.g., [1, 2])
    result = []
    
    # Helper function, current_subset keeps track of the path we are currently building
    def backtrack(index, current_subset):
        # Base case - if we have made a decision for all elements, save the subset
        if index == len(nums):
            result.append(current_subset.copy()) # Append a copy to prevent mutation
            return
        
        # Decision 1 - include nums[index]
        current_subset.append(nums[index])
        # Moves forward to make a decision about the next number, while carrying the number we just added.
        backtrack(index + 1, current_subset)
        
        # Backtrack - undo the inclusion choice
        current_subset.pop() # This undoes Decision 1. 
        # We remove the number we just added so we can explore what happens when we don't include it.
        
        # Decision 2: Exclude nums[index]
        # We move forward to make a decision about the next number, this time excluding the current number.
        backtrack(index + 1, current_subset)
    
    # Kicks off the backtracking process.
    backtrack(0, [])
    # Returns the final list of all possible subsets after all branches are explored.
    return result
```

**How it executes step-by-step**:

Given nums = [1, 2]:
- backtrack(0, []) updates subset to [1] => calls backtrack(1, [1]).
- backtrack(1, [1]) updates subset to [1, 2] => calls backtrack(2, [1, 2]).
    - Base case reached: [1, 2] is saved.
- Backtracks (pops 2). Current subset is [1].
- Excludes 2 => calls backtrack(2, [1]).Base case reached: [1] is saved.
- Backtracks (pops 1). Process continues similarly for exclusion paths, resulting in [] and [2].

## Dynamic Programming

**Problem**: Climbing Stairs (LeetCode 70)

**Strategy**: Recognize overlapping subproblems. To reach step n, you must either step from n-1 or n-2. Store subproblem results in a tracking table (or variables) to optimize from an exponential time runtime down to a single linear pass.

```python
# Bottom-Up Dynamic Programming (Tabulation approach)
def climbStairs(n):
    # Handle base constraints directly
    if n <= 2:
        return n
    
    # Initialize DP table array to store solutions to subproblems
    dp = [0] * (n + 1) # O(n)
    dp[1] = 1 # 1 way to reach the 1st step
    dp[2] = 2 # 2 ways to reach the 2nd step (1+1 or 2)
    
    # Fill out the table iteratively using the recurrence relation
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    
    return dp[n]
```

```python
# Space-Optimized Variation: O(1) Space
def climbStairsOptimized(n):
    if n <= 2:
        return n
    
    first, second = 1, 2
    
    for _ in range(3, n + 1):
        first, second = second, first + second
    
    return second
```

**How it executes step-by-step**:

Given n = 4:
- The code sets up an empty state table array size 5: [0, 1, 2, 0, 0].
- Loops starting from index 3:
    - For Step 3: dp[3] = dp[2] + dp[1] => 2 + 1 = 3. Table becomes [0, 1, 2, 3, 0].
    - For Step 4: dp[4] = dp[3] + dp[2] => 3 + 2 = 5. Table becomes [0, 1, 2, 3, 5].
- Returns dp[4], which equals 5.

### Top-Down Memoization alternative for Dynamic Programming

*Top-Down Memoization* optimizes a naive recursive solution by caching the results of expensive function calls in a hash map or array. When the algorithm encounters a previously calculated state, it returns the cached result instantly instead of recalculating it.

**The Pattern Template**

A standard Top-Down Dynamic Programming solution always follows these three structural steps:
- **Check the cache**: Return the saved result immediately if the current state exists in your memo dictionary.
- **Handle base cases**: Return hardcoded values for the smallest possible subproblems.
- **Compute and store**: Calculate the result recursively, save it to the cache, and then return it.

**Problem**: Climbing Stairs (LeetCode 70)

```python
def climbStairs(n):
    # Initialize a memoization cache (dictionary)
    memo = {}
    
    def total_ways(current_step):
        # Step 1: Check the cache to see if we already know the answer
        if current_step in memo:
            return memo[current_step]
        
        # Step 2: Handle base cases
        if current_step == 1:
            return 1
        if current_step == 2:
            return 2
        
        # Step 3: Compute recursively, store in the cache, and return
        memo[current_step] = total_ways(current_step - 1) + total_ways(current_step - 2)
        return memo[current_step]
    
    return total_ways(n)
```

How It Executes Step-by-Step (n = 4)
- total_ways(4) runs. It is not in memo. It calls total_ways(3) and total_ways(2).
- total_ways(3) runs. It is not in memo. It calls total_ways(2) and total_ways(1).
- total_ways(2) hits a base case and returns 2.
- total_ways(1) hits a base case and returns 1.
- total_ways(3) sums those results (2 + 1 = 3), saves {3: 3} to memo, and returns 3.
- The execution goes back to total_ways(4), which now evaluates its second call: total_ways(2).
- total_ways(2) immediately hits the base case criteria and returns 2.
- total_ways(4) sums the results (3 + 2 = 5), saves {4: 5} to memo, and returns 5.

**Python Pro-Tip**: The @lru_cache Decorator

In coding interviews, I can completely automate the memoization wrapper step using Python's built-in functools module. This keeps your code exceptionally clean.

```python
from functools import lru_cache

class Solution:
    def climbStairs(self, n: int) -> int:
        
        # maxsize=None transforms the recursive function into a memoized one
        @lru_cache(maxsize=None)
        def total_ways(current_step: int) -> int:
            if current_step <= 2:
                return current_step
                
            return total_ways(current_step - 1) + total_ways(current_step - 2)
            
        return total_ways(n)
```

**Summary**: Top-Down vs. Bottom-Up
- **Top-Down (Memoization)** starts at the final target goal (n) and breaks it down backward. It is often easier to conceptualize because it mirrors a standard recursive path.
- **Bottom-Up (Tabulation)** starts at the base cases (1 and 2) and builds forward using an iterative loop. It is generally more memory efficient because it eliminates recursion stack overhead.

## Tips

**Integer Division**: Always use *//* for floor division. Standard */* converts values to a float.

**Infinity Bounds**: Use *float('inf')* and *float('-inf')* to initialize minimum and maximum trackers seamlessly.

**Efficient Queues**: Never use a standard Python list as a queue; dropping elements from the front *(list.pop(0))* takes *O(N)* time. Always import from collections import deque to utilize *.popleft()*, which runs in *O(1)* time.
