## Q2. What’s the difference between a list, tuple, and set?
### Answer:
| Type  | Mutable | Ordered | Duplicates |
| ----- | ------- | ------- | ---------- |
| list  | ✅ Yes   | ✅ Yes   | ✅ Yes      |
| tuple | ❌ No    | ✅ Yes   | ✅ Yes      |
| set   | ✅ Yes   | ❌ No    | ❌ No       |

```python
nums = [1, 2, 3]       # list [mutable]
coords = (1, 2, 3)     # tuple (immutable)
unique = {1, 2, 2, 3}  # set → {1, 2, 3}
print(unique)
```

## Q3. Explain shallow vs. deep copy.
### Answer:
- Shallow copy: copies references (changes affect both).
- Deep copy: creates an entirely new object.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
c = copy.deepcopy(a)

a[0][0] = 99
print(b)  # [[99, 2], [3, 4]] → affected
print(c)  # [[1, 2], [3, 4]] → unaffected
```

## Q4. What is a Python decorator?
### Answer:
A decorator is a function that modifies another function’s behavior.

```python
def logger(fn):
    def wrapper():
        print("Before call")
        fn()
        print("After call")
    return wrapper

@logger
def greet():
    print("Hello!")

greet()
```

## Q5. What are list comprehensions?
### Answer:
A concise way to create lists.

```python
nums = [1, 2, 3, 4]
squares = [n**2 for n in nums if n % 2 == 0]
print(squares)  # [4, 16]
```
