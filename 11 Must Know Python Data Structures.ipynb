# Must know data structures

## Lists (Dynamic Arrays)

Python lists are mutable, ordered sequences. They act as your go-to structure for arrays, matrix grids, sliding windows, and two-pointer interview patterns.

append(), pop(); Time Complexity O(1)

```python
# Initialization
arr = [1, 2, 3]
```

```python
# Always use list comprehension for 2D arrays to avoid shallow copy bugs
matrix = [[0]*3 for _ in range(3)]
```

```python
# Append O(1)
arr.append(4)
print(arr)
```

```python
# Pop from end 
arr.pop()
print(arr)
```

```python
# Insert/delete by index - O(n)
arr.insert(0, 5)
print(arr)
```

```python
val = arr[2]
print(val)
```

```python
# O(n)
arr.pop(0)
print(arr)
```

```python
# Slicing O(k)
# arr[start:end:step]
print(arr[:2])
```

**Note**: Avoid using arr.pop(0) in a loop. Shifting elements makes this operation an expensive O(N) mistake, turning your O(N) algorithm into an inefficient O(N²) nightmare.

## Dictionaries (Hash Maps)

Dictionaries store key-value pairs with average O(1) insertion and lookup speeds. They are indispensable for counting element frequencies, indexing data, and memoization in dynamic programming.

defaultdict, get(); O(1) average

```python
# Initialization
hash_map = {}
from collections import defaultdict
d = defaultdict(int)
print(d)
```

```python
# Get with default - Prevents KeyError exceptions when a key is missing
hash_map.get("key", 0)
```

```python
# Iterate
for key, val in hash_map.items():
    print("loops through both components simultaneously")
```

**The defaultdict Advantage:** 

Using collections.defaultdict(list) automatically initializes missing keys with an empty list, eliminating noisy boilerplates like if key not in d:.

```python
from collections import defaultdict

counts = defaultdict(int)
for char in "interview":
    counts[char] += 1

if "i" in counts:
    print(counts)
    print("Counts of i:", counts["i"])
```

## Sets (Hash Sets)

Sets are unordered collections of unique, hashable elements. Use them for O(1) membership testing, rapid deduplication, and tracking visited nodes in graph traversals.

add(), in; O(1)

```python
# Initialization
# Note: {} creates an empty dictionary, not a set
unique = set()
```

```python
unique.add(1)
print(unique)
```

```python
unique.remove(1)
print(unique)
```

**Membership check:** if val in unique: — O(1) time complexity, compared to O(N) for lists.

```python
seen = set()

# Core operations
seen.add(5) # O(1)
if 5 in seen:
    print("Found")

# Set operations 
set_a, set_b = {1, 2}, {2, 3}
inter_section = set_a & set_b 
print(inter_section)
```

## Deques (Double-Ended Queues)

While a regular list works well as a LIFO stack, it makes a terrible FIFO queue. 

Python's collections.deque provides fast O(1) appends and pops from both ends, making it mandatory for Breadth-First Search (BFS) graph traversals.

append(), popleft(); O(1)

```python
from collections import deque

queue = deque([1, 2, 3])

# Queue behavior FIFO
queue.append(4)              # Add to back
first_item = queue.popleft() # O(1) - Remove from front
print(first_item)
```

```python
# Stack behavior LIFO
queue.appendleft(0)      # O(1) - Add to front
last_item = queue.pop() # O(1) - Remove from back
print(last_item)
```

## Heaps (Priority Queues)

heappush(), heappop(); O(log N)

Heaps are binary trees that allow you to retrieve the minimum or maximum element in O(1) time while handling insertions in O(log N) time. Use them for "Top K" elements, stream processing, or scheduling algorithms.

**Note**: Python’s heapq module only implements a Min-Heap out of the box. 

**Max-Heap Workaround**: Multiply your values by -1 before pushing them into the heap to simulate a max-heap. When you pop a value, multiply it by -1 again to restore its original positive state.

```python
import heapq
min_heap = []

# Push elements - O(logn)
heapq.heappush(min_heap, 10)
heapq.heappush(min_heap, 5)
heapq.heappush(min_heap, 20)

# Peak smallest element - O(1)
smallest = min_heap[0]
print(smallest)
```

```python
# Pop smallest element - O(logn)
print(heapq.heappop(min_heap))
```

```python
# Convert an existing list to a heap in-place in O(N) time
nums = [5, 7, 9, 1]
heapq.heapify(nums)
print(nums[0])
```

## Linked Lists

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

# Creating a simple list: 1->2
head = ListNode(1)
head.next = ListNode(2)

print(head.val)
print(head.next)
```

```python
# Traversal pattern
curr = head
while curr:
    print(curr.val)
    curr = curr.next
```

## Binary Trees

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# Creating a simple tree
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
print(root.val)
```
