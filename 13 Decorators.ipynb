## Decorators

The original function stays focused on its job. The decorator handles additional behavior. Here is an example:

```python
from functools import wraps
```

```python
def require_admin(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        # Extract the role from kwargs safely
        user_role = kwargs.get("role")
        
        # Check the value of the role
        if user_role == "admin":
            return func(*args, **kwargs)
        
        # Handle unauthorized access
        raise PermissionError("Access denied: Admin role required")
        
    return wrapper
```

```python
@require_admin
def delete_user(user, role="user"):
    print(f"Deleting {user}")
```

```python
delete_user("Odysseus", role="admin")
```

```python
try:
    delete_user("Odysseus", role="user")
except PermissionError:
    print("Raises Permission Error: Access denied: Admin role required")
```
