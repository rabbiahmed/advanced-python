## Safe Resource Management

Python guarantees that the file gets closed when the "with" block exits—even if an exception occurs. "with" is Python's mechanism for guaranteeing resource lifecycle management.

```python
from contextlib import contextmanager
```

```python
@contextmanager
def transaction():
    print("BEGIN TRANSACTION")
    
    try:
        yield
    except Exception:
        print("ROLLBACK TRANSACTION")
        raise
    else:
        print("COMMIT TRANSACTION")
    finally:
        print("CLOSE CONNECTION")
```

```python
with transaction():
    print("Updating database...")
```

```python
with transaction():
    print("Updating database...")
    # Re-Raise the exception instead suppress the exception. Let the caller/application handle it.
    raise ValueError("Something went wrong")  
```
