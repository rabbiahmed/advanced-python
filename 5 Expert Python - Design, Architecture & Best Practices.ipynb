## Q31. What are the core principles of OOP in Python?
### Answer:
- Encapsulation – hiding internal data
- Abstraction – exposing only essential parts
- Inheritance – reusing code from parent classes
- Polymorphism – same interface, different behavior

```python
# Use OOP when data + behavior belong together.

class Animal:
    def speak(self): 
        pass

class Dog(Animal):
    def speak(self): 
        return "Woof!"

class Cat(Animal):
    def speak(self): 
        return "Meow!"

for pet in [Dog(), Cat()]:
    print(pet.speak())  # polymorphism
```

## Q32. Explain composition vs inheritance.
### Answer:
- Inheritance: “is-a” relationship.
- Composition: “has-a” relationship — more flexible and modular.

```python
# Use composition when I want reusable components without deep hierarchies.

class Engine:
    def start(self): 
        print("Engine is starting")

class Car:
    def __init__(self):
        self.engine = Engine()  # composition
    def drive(self):
        self.engine.start()
        print("Car is moving")
```

## Q33. What are Python design patterns you should know?
### Commonly used patterns:
| Pattern   | Type       | Example                               |
| --------- | ---------- | ------------------------------------- |
| Singleton | Creational | Logger, Config Manager                |
| Factory   | Creational | Object creation with abstraction      |
| Observer  | Behavioral | Event system                          |
| Strategy  | Behavioral | Switch between algorithms dynamically |
| Decorator | Structural | Extend behavior dynamically           |
| Facade    | Structural | Provides a simplified interface       |

```python
# Example: Singleton Pattern
# Override the __new__ method like this:

class Singleton(object):
    _instance = None
    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super(Singleton, cls).__new__(cls, *args, **kwargs)
        return cls._instance

if __name__ == '__main__':
    s1 = Singleton()
    s2 = Singleton()
    
    if (id(s1) == id(s2)):
        print("Same")
    else:
        print("Different")
```

```python
# Example: Factory Pattern

class Shape:
    def draw(self): 
        pass

class Circle(Shape):
    def draw(self): 
        print("Circle")

class Square(Shape):
    def draw(self): 
        print("Square")

def shape_factory(shape_type):
    if shape_type == "circle": 
        return Circle()
    if shape_type == "square": 
        return Square()
```

## Q34. What is dependency injection?
### Answer:
It means passing dependencies (objects or functions) into a class, rather than hardcoding them. Makes code more testable and decoupled.

```python
class NotificationService:
    def __init__(self, sender):
        self.sender = sender
    
    def notify(self, msg):
        self.sender.send(msg)

class EmailSender:
    def send(self, msg):
        print("Email:", msg)

# Inject EmailSender dependency
service = NotificationService(EmailSender())
service.notify("Hello!")
```

## Q36. What’s the difference between duck typing and static typing?
### Answer:
- Duck typing: "If it walks like a duck and quacks like a duck..." → Focuses on behavior, not type.
- Static typing: enforces explicit type declarations at compile time.

```python
class Duck:
    def quack(self): 
        print("Quack!")

class Person:
    def quack(self): 
        print("I'm pretending to be a duck")

def make_it_quack(thing):
    thing.quack()   # Works as long as method exists

make_it_quack(Person())
```

## Q37. What are functional programming tools in Python?
### Answer:
Python supports hybrid OOP + functional paradigms.

| Tool                     | Description                              |
| ------------------------ | ---------------------------------------- |
| `map(func, iterable)`    | Applies func to each element             |
| `filter(func, iterable)` | Filters elements where func returns True |
| `reduce(func, iterable)` | Aggregates values cumulatively           |
| `functools.partial()`    | Pre-fills arguments                      |
| `functools.lru_cache()`  | Memoizes function results                |

```python
from functools import reduce, lru_cache

nums = [1,2,3,4]
print(list(map(lambda x: x*2, nums)))      # [2,4,6,8]
print(list(filter(lambda x: x%2==0, nums))) # [2,4]

@lru_cache(maxsize=None)
def fib(n): 
    return n if n<2 else fib(n-1) + fib(n-2)
```

## Q39. How do you ensure thread safety in Python?
### Answer:
Use synchronization primitives from threading.

Other options:
- RLock, Semaphore, Event
- Queue.Queue (thread-safe data structure)

```python
import threading

lock = threading.Lock()
counter = 0

def safe_increment():
    global counter
    with lock:
        counter += 1
```
