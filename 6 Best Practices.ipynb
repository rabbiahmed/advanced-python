## Q49. How do you use dataclasses for immutability and defaults?
### Answer:
- frozen=True makes objects immutable.
- default_factory avoids mutable default traps.

```python
from dataclasses import dataclass, field

@dataclass(frozen=True)
class Config:
    name: str
    values: list = field(default_factory=list)
```

## Q50. What are the most common real-world Python pitfalls?
| Pitfalls                    | Example                                 | Fix                              |
| --------------------------- | --------------------------------------- | -------------------------------- |
| Mutable default args        | `def f(x=[])`                           | Use `None` + init inside         |
| GIL misunderstanding        | Threads for CPU                         | Use `multiprocessing`            |
| Late binding in lambdas     | `funcs = [lambda: i for i in range(3)]` | Capture default: `lambda i=i: i` |
| Circular imports            | Modules import each other               | Refactor or use local import     |
| Silent exception swallowing | `try/except: pass`                      | Always log or handle properly    |

```python
# Problem
def add_item(item, my_list=[]):
    my_list.append(item)
    return my_list

print(add_item(1))  # Output: [1]
print(add_item(2))  # Output: [1, 2] - Unexpected!
print(add_item(3))  # Output: [1, 2, 3] - Still unexpected!
```

```python
# Solution
def add_item_corrected(item, my_list=None):
    if my_list is None:
        my_list = []  # Initialize a new list for each call
    my_list.append(item)
    return my_list

print(add_item_corrected(1))  # Output: [1]
print(add_item_corrected(2))  # Output: [2] - Correct!
print(add_item_corrected(3))  # Output: [3] - Correct!
```

```python
# Porblem
funcs = [lambda: i for i in range(3)]
for f in funcs:
    print(f())
```

```python
# Solution - Late binding with default arguments
# In this case, lambda i=i: i defines a lambda function where i is both a parameter and its default value is set to the 
# current value of the outer i at the time of the lambda's creation. 
# This ensures that each lambda captures a distinct value, resolving the late binding issue.
funcs = [lambda i=i: i for i in range(3)]
for f in funcs:
    print(f())
```
