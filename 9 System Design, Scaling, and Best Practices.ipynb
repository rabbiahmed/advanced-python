## Q1. What are Python’s common design patterns?
### Answer:
Design patterns are reusable solutions to common problems in software design.

| Pattern       | Description                            | Example                |
| ------------- | -------------------------------------- | ---------------------- |
| **Singleton** | Only one instance exists               | Logging system         |
| **Factory**   | Creates objects without exposing logic | Object creation API    |
| **Observer**  | Notifies dependents of state change    | Event-driven systems   |
| **Decorator** | Adds behavior dynamically              | Flask route decorators |

```python
# Singleton using class attribute
class Singleton:
    _instance = None
    def __new__(cls):
        if not cls._instance:
            cls._instance = super().__new__(cls)
        return cls._instance

s1, s2 = Singleton(), Singleton()
print(s1 is s2)  # True
```

## Q3. How can you improve Python performance?
### Answer:
1. Code-Level Optimizations
- Use built-in functions (C-optimized): sum(), min(), etc.
- Prefer list comprehensions over manual loops.
- Avoid excessive object creation.
- Use generators for streaming data.

2. System-Level Optimizations
- Use multiprocessing for CPU-bound tasks.
- Use asyncio or threading for I/O-bound tasks.
- Profile with cProfile and timeit.

```python
import timeit
print(timeit.timeit('[x*x for x in range(1000)]', number=1000))
```

## Q4. What are memory leaks in Python, and how to detect them?
### Answer:
Memory leaks happen when references to unused objects persist.

### Common Causes:
- Global variables never cleared
- Circular references
- Unclosed file handles

### Tools:
- gc module → gc.collect(), gc.get_objects()
- objgraph library → visualize references
- memory_profiler for line-by-line memory tracking

from memory_profiler import profile

@profile

def leaky():

    leak = []
    for _ in range(10**5):
        leak.append(object())

## Q5. How do you handle concurrency safely in Python?
### Answer:
Python’s GIL means only one thread runs Python bytecode at a time.
So:

| Type                | When to Use          | Tools                                                |
| ------------------- | -------------------- | ---------------------------------------------------- |
| **Threading**       | I/O-bound tasks      | `threading`, `concurrent.futures.ThreadPoolExecutor` |
| **Multiprocessing** | CPU-bound tasks      | `multiprocessing`, `ProcessPoolExecutor`             |
| **Async I/O**       | High I/O concurrency | `asyncio`, `aiohttp`, `trio`                         |

```python
import concurrent.futures

def square(x):
    return x*x

with concurrent.futures.ThreadPoolExecutor() as executor:
    results = executor.map(square, range(5))
    print(list(results))
```
