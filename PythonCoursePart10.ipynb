# Putting It All Together

## 1. Core Concepts
| Concept              | Summary                                         | Example                            |
| -------------------- | ----------------------------------------------- | ---------------------------------- |
| Dynamic Typing       | Variables can change type at runtime            | `x=5; x='Hi'`                      |
| Mutable vs Immutable | Lists, dicts mutable; tuples, strings immutable | `a=[1]; b=a; a.append(2)`          |
| Shallow vs Deep Copy | Shallow copies refs, deep copies objects        | `copy.copy()` vs `copy.deepcopy()` |
| List Comprehension   | Compact loops                                   | `[x*x for x in nums if x>0]`       |
| Decorators           | Modify function behavior                        | `@logger`                          |
| `is` vs `==`         | `is` → same object; `==` → equal value          | `a is b`, `a == c`                 |

## 2. Intermediate Python
| Concept               | Description                                    | Example              |
| --------------------- | ---------------------------------------------- | -------------------- |
| `*args`, `**kwargs`   | Variable args & keyword args                   | `def f(a,*b,**c)`    |
| Generator             | Lazy iterator via `yield`                      | `yield i`            |
| Module vs Package     | File vs Folder with `__init__.py`              | `import pkg.mod`     |
| Context Manager       | Auto-manages resources                         | `with open() as f:`  |
| Exception Handling    | `try/except/else/finally`                      | `except ValueError:` |
| Iterables & Iterators | Iterable: `__iter__()`; Iterator: `__next__()` | `iter(list)`         |

## 3. Modern Python (3.6–3.12)
| Feature          | Purpose                        | Example                      |
| ---------------- | ------------------------------ | ---------------------------- |
| f-Strings        | Fast string formatting         | `f"{name} {age}"`            |
| Dataclasses      | Auto-init classes              | `@dataclass class P:`        |
| Type Hints       | Static typing                  | `def add(a:int,b:int)->int:` |
| Async/Await      | Non-blocking I/O               | `await asyncio.sleep(1)`     |
| Pattern Matching | Structural switch-case (3.10+) | `match x:`                   |
| Walrus Operator  | Inline assignment              | `if (n:=len(a))>3:`          |

## 4. Concurrency & Performance
| Technique       | For                              | Library                           |
| --------------- | -------------------------------- | --------------------------------- |
| Threading       | I/O-bound                        | `threading`, `concurrent.futures` |
| Multiprocessing | CPU-bound                        | `multiprocessing`                 |
| Async I/O       | Many concurrent tasks            | `asyncio`, `aiohttp`              |
| Caching         | Speed reuse                      | `functools.lru_cache`             |
| Profiling       | Find bottlenecks                 | `cProfile`, `timeit`              |
| GIL             | One thread runs bytecode at once | Use multiprocessing to bypass     |

## 5. Architecture, Testing & Deployment
| Topic                 | Key Points                                         |
| --------------------- | -------------------------------------------------- |
| **Design Patterns**   | Singleton, Factory, Observer, Decorator            |
| **Project Structure** | `app/`, `tests/`, `requirements.txt`, `Dockerfile` |
| **Testing**           | Use `pytest`, `unittest.mock`, `pytest --cov`      |
| **Static Analysis**   | `flake8`, `mypy`, `black`                          |
| **CI/CD**             | GitHub Actions / Jenkins pipelines                 |
| **Deploy**            | Docker + Gunicorn + Nginx or AWS Lambda            |
| **Memory Leaks**      | Detect via `gc`, `objgraph`, `memory_profiler`     |

## 6. Best Practices
- Use venv / Docker for isolation
- Handle exceptions narrowly
- Avoid mutable defaults (def f(a=[]))
- Prefer comprehensions & built-ins
- Use logging, not print()
- Follow PEP 8 & type hints
- Write docstrings
- Modularize for readability

## 7. Common Short Questions
| Q                                       | A                                            |
| --------------------------------------- | -------------------------------------------- |
| Why is Python not fully parallel?       | Global Interpreter Lock (GIL)                |
| Difference between deep & shallow copy? | Deep copies objects recursively              |
| How to release memory?                  | Delete refs + `gc.collect()`                 |
| How to speed up Python?                 | Use built-ins, async, C extensions           |
| Virtualenv purpose?                     | Dependency isolation                         |
| Mutable vs Immutable?                   | Mutable → list, dict; Immutable → str, tuple |

## 9. Final Tips
- Always explain why you choose a structure or library.
- Discuss trade-offs (readability vs performance).
- Mention testing, type hints, and asynchronous design when possible.
- Write clean, modular, and PEP-8 compliant code.
- Use examples that show understanding of real-world scaling.
