## Functions as First-Class Objects

We want to avoid long if/else chains. It's better to replace those with dispatch tables in production. Treat functions like data. So, adding new behavior means addin a fucntion, not editing existing logic. This also follows the Open/Closed principle. 
- Open for extension
- Closed for modification

```python
def start():
    print("Starting server...")

def stop():
    print("Stopping server...")

def restart():
    print("Restarting server...")

def status():
    print("Server is running...")
```

```python
COMMANDS = {'start': start, 'stop': stop, 'restart': restart, 'status': status}

def execute(command):
    handler = COMMANDS.get(command)
    
    if handler is None:
        print(f"Unknown command: {command}")
        return
    
    handler()
```

```python
# Example
execute("start")
execute("status")
execute("restart")
execute("stop")
execute("invalid")
```

### Better Pythonic version

```python
def start(server):
    print(f"Starting {server}...")

def stop(server):
    print(f"Stopping {server}...")

def restart(server):
    print(f"Restarting {server}...")

def status(server):
    print(f"{server} is running...")

COMMANDS = {
    'start': start, 
    'stop': stop, 
    'restart': restart, 
    'status': status,
}

def unknown(server):
    print(f"Unknown command for {server}")

def execute(command, server):
    COMMANDS.get(command, unknown)(server) # .get() avoids a keyError and handles unknown inputs gracefully
```

```python
execute("start", "web-server")
execute("status", "web-server")
execute("restart", "web-server")
execute("stop", "web-server")
execute("invalid", "web-server")
```
