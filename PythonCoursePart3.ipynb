### Answer:
Manages resources like files automatically using with.

```python
with open("data.txt", "w") as f:
    f.write("Hello!")
# File automatically closed
```

## Q14. What are dataclasses?
### Answer:
Introduced in Python 3.7 — automatically adds __init__, __repr__, etc.

In Python, __init__ and __repr__ are special methods (also known as "magic methods" or "dunder methods") that define how objects of a class behave in specific situations.

The __init__ method is the constructor or initializer for a class. 
- It is automatically called when a new instance of a class is created.
- Its primary purpose is to initialize the attributes (data) of the newly created object.
- It takes self as its first argument, which refers to the instance being created, and can accept other arguments to set the initial state of the object.

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

my_dog = Dog("Buddy", "Golden Retriever") # __init__ is called here
print(my_dog.name) # Output: Buddy
```

The __repr__ method defines the official string representation of an object. 
- It is called by the built-in repr() function and is typically used for debugging and development.
- The goal of __repr__ is to return a string that, if passed to eval(), would ideally recreate the object (or at least provide a clear, unambiguous representation of its state).
- It should aim to be as unambiguous as possible and often includes the class name and the values of key attributes.

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

p = Point(1, 2)
print(p)  # Point(x=1, y=2)
```

## Q15. How do you handle exceptions properly?
### Answer:
Use try/except/finally, optionally with else.

```python
try:
    x = 1 / 0
except ZeroDivisionError as e:
    print("Error:", e)
else:
    print("No error")
finally:
    print("Always runs")
```
