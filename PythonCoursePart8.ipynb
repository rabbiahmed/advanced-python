## Q61. What are idempotent APIs and why do they matter?
### Answer:
Idempotent: multiple identical requests ⇒ same result. Important for retries and fault tolerance.

Example: PUT /users/1 with same payload → no duplicate users.

## Q62. How do you secure a Python web service?
### Answer:
1. Use HTTPS / TLS
2. Sanitize inputs (pydantic validation)
3. Store passwords hashed (bcrypt)
4. Use JWT or OAuth2 for auth
5. Rate-limit and log suspicious behavior

from passlib.hash import bcrypt

hashed = bcrypt.hash("secret")

assert bcrypt.verify("secret", hashed)

## Q64. Describe how you’d build a microservice ecosystem in Python.
### Answer:
Each microservice:
- Has its own DB + REST/GraphQL API
- Communicates via message queue
- Shares schema through contracts (e.g., protobuf)
- Uses CI/CD pipelines and containers

Common tools: FastAPI, Celery, Redis, Docker, Kubernetes, NGINX, Prometheus

## Q65. How do you containerize and deploy Python apps?

### Dockerfile Example
FROM python:3.12-slim

WORKDIR /app

COPY requirements.txt .

RUN pip install -r requirements.txt

COPY . .

CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]

#### Then:

docker build -t myapp .

docker run -p 8000:8000 myapp

Add health checks + readiness probes when orchestrating with Kubernetes.

## Q68. How do you handle configuration and secrets in production?
- Environment variables via dotenv or pydantic-settings
- Secret managers: AWS SM, Azure KeyVault, HashiCorp Vault
- Never commit .env or API keys

## Q69. What’s the role of asynchronous frameworks like FastAPI?
- They use async/await to handle thousands of simultaneous connections efficiently (non-blocking).
- Ideal for real-time APIs, chat servers, or IoT backends.

## Q70. How do you scale a Python system horizontally?
- Stateless containers
- Shared cache (Redis) and DB
- Load balancer (NGINX / AWS ALB)
- Message queues for async work
- Use auto-scaling groups or Kubernetes HPA
