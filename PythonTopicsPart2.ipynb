## Q6. Explain args and kwargs.
### Answer:
Used to pass variable-length arguments.

```python
def show(a, *args, **kwargs):
    print(a, args, kwargs)

show(1, 2, 3, x=10, y=20)
# a=1, args=(2,3), kwargs={'x':10, 'y':20}
```

## Q7. What are generators and how do they differ from lists?
### Answer:
Generators use yield to produce items lazily (one by one).
#### Advantage: 
Memory efficient (doesn’t store the entire sequence).

```python
def gen_numbers():
    for i in range(3):
        yield i

g = gen_numbers()
print(next(g))  # 0
print(next(g))  # 1
print(next(g))  # 2
```

## Q8. Explain Python’s memory management.
### Answer:
- Managed by Python’s garbage collector.
- Uses reference counting + cyclic GC (generation 1, 2, 3).
- Use del to remove object references.

## Q9. What is the difference between is and ==?
### Answer:
- is → compares object identity
- == → compares values

```python
a = [1,2,3]
b = a
c = [1,2,3]

print(a is b)  # True
print(a == c)  # True
print(a is c)  # False
```
